# Research disciplines of the papers
Which fields do MBDef's papers come from? MBDef records each paper's DOI and OpenAlex id but no discipline, so this notebook asks [OpenAlex](https://help.openalex.org/data/topics/) for each paper's **primary topic** and counts papers per field, seed papers in grey and closure papers in blue. An inset breaks Social Sciences down into its subfields. Medicine and Health Professions are drawn as one bar; the saved tables keep them apart.

OpenAlex's topics are automatic classifications, not checked discipline labels. No language model is used.

## Load MBDef
`load()` reads MBDef's five tables from `$DATA_ROOT/MBDef/data/`, where `uv run mbdef data fetch MBDef/` put them.

```python
from pathlib import Path

from IPython.display import SVG, display

from mbdef.analysis import load

OUTPUTS = Path("outputs")  # tutorials/outputs/, not tracked by git
data = load()
```

## Ask OpenAlex
`run` looks every paper up by its OpenAlex id (50 per request), then by DOI the papers that this did not find, and keeps every answer in `outputs/research-disciplines/openalex-cache.json`. A run that stops (network, rate limit) resumes from the cache when you run the cell again, and a second run asks nothing. With `OFFLINE = True` it reads only the cache. An `OPENALEX_API_KEY` in the repository's `.env` raises OpenAlex's daily budget; it is optional.

OpenAlex reclassifies works over time, so a fresh cache can give slightly different counts. Keep the cache with any numbers you publish.

```python
from mbdef.analysis.disciplines import run

OFFLINE = False
report = run(data, OUTPUTS / "research-disciplines", offline=OFFLINE)
print("Papers:", report["papers"])
print("Identifiers:", report["identifier_coverage"])
print("Status:", report["status"])
```

## Fields and the figure
Each paper counts once, under the field of its primary topic. The bars add up to every paper: the ten largest fields, the remaining classified fields, and the papers OpenAlex did not classify. A label `total (seed, closure)` stands outside a bar when the counts do not fit inside it.

```python
for field in report["counts"]["primary"]["field"][:10]:
    print(
        f"{field['name']:<40} {field['papers']:>5}  {field['percent_of_corpus']:5.1f}%  {field['by_corpus']}"
    )
display(SVG(filename=OUTPUTS / "research-disciplines" / "research-disciplines.svg"))
```
